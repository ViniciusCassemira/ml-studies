## Exemplo de uso TF-IDF

### Importando bibliotecas que serão utilizadas

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

# Opções de exibição do pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 10000)
pd.set_option('display.precision', 5)
```

```python
# Declarações dos exemplos
corpus = [
    "o gato sentou no tapete",
    "o cachorro brincou no parque",
    "gato e cachorro são bons animais"
]
consulta = "o gato brincou no quintal"

print("# Corpus:")
for index ,i in enumerate(corpus):
    print(f"  Documento {index + 1}: '{i}'")
print()
print(f"# Consulta: '{consulta}'")
print()

# Criando objeto TfidfVectorizer
vectorizer = TfidfVectorizer()
# Gerando a matriz TF-IDF para o corpus
corpus_tfidf_matrix = vectorizer.fit_transform(corpus)

# Utilizamos o pandas DataFrame para exibir a matriz TF-IDF de forma mais legível
# Nota: As stop_wowrds foram removidas automaticamente, então palavras como "o", "no", "e" não aparecem na matriz
print("# Matriz TF-IDF: (sem 'e' e 'o')")
corpus_df_tfidf = pd.DataFrame(
    corpus_tfidf_matrix.toarray(), 
    columns=vectorizer.get_feature_names_out(),
    index=[f'Documento {i+1}' for i in range(len(corpus))]
)
print(corpus_df_tfidf)
print()

# Gerando a matriz TF-IDF para a consulta com base no vocabulário do corpus
print("# Matriz TF-IDF com a consulta: (sem 'e' e 'o')")
consulta_tfidf_matrix = vectorizer.transform([consulta])
consulta_df_tfidf = pd.DataFrame(
    consulta_tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f'Consulta' for i in range(len([consulta]))]
)
# Concatenando a matriz TF-IDF do corpus com a matriz TF-IDF da consulta
print(pd.concat([
    corpus_df_tfidf,
    consulta_df_tfidf
]))
print()

# Calculando a similaridade entre a consulta e os documentos do corpus
similaridade = cosine_similarity(consulta_tfidf_matrix, corpus_tfidf_matrix)[0]

# Ranking da similaridade (do maior para o menor) entre a consulta e os documentos do corpus
print("# Similaridade entre consulta e corpus:")
ranking = similaridade.argsort()[::-1]
for i in ranking:
    print(f"  Documento {i + 1}: {similaridade[i]}")
```
